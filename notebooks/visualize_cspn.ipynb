```python
import torchvision.utils as vutils
import matplotlib.pyplot as plt


def show(tensor, title=None):
    grid = vutils.make_grid(tensor.cpu(), nrow=8, normalize=True)
    plt.figure(figsize=(16, 4))
    plt.imshow(grid.permute(1, 2, 0))
    plt.axis("off")
    if title:
        plt.title(title)
    plt.show()
```
