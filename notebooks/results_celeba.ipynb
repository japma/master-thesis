```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

plt.rcParams["figure.dpi"] = 200

RESULTS = Path("../results/celeba")
METRICS = {"fid": "lower is better", "kid": "lower is better", "cmmd": "lower is better",
           "precision": "higher is better", "recall": "higher is better"}
```

```python
results = {metric: pd.read_csv(RESULTS / f"{metric}.csv") for metric in METRICS}
results["fid"].groupby(["checkpoint", "source"])["value"].agg(["mean", "std", "count"])
```

```python
def plot_metric(metric):
    df = results[metric]
    generated = df[df["source"] == "generated"].groupby("checkpoint")["value"].agg(["mean", "std"])

    fig, ax = plt.subplots(figsize=(8, 5))
    bars = ax.bar(generated.index, generated["mean"], yerr=generated["std"], capsize=4, color="tab:blue")
    ax.bar_label(bars, fmt="{:.4g}", padding=3, fontsize="small")

    for source, colour in [("real", "tab:green"), ("reconstruction", "tab:orange")]:
        reference = df[df["source"] == source]["value"].mean()
        ax.axhline(reference, color=colour, linestyle="--", label=f"{source} ({reference:.4g})")

    ax.set_title(f"{metric} ({METRICS[metric]})")
    ax.tick_params(axis="x", labelrotation=45)
    plt.setp(ax.get_xticklabels(), ha="right")
    ax.legend(fontsize="small")
    fig.tight_layout()
    plt.show()
```

```python
plot_metric("fid")
```

```python
plot_metric("kid")
```

```python
plot_metric("cmmd")
```

```python
plot_metric("precision")
```

```python
plot_metric("recall")
```
