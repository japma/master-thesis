```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

plt.rcParams["figure.dpi"] = 200

RESULTS = Path("../results/colourmnist")
```

```python
accuracy = pd.read_csv(RESULTS / "accuracy.csv")
accuracy.fillna({"vae": "-", "checkpoint": "-"}).pivot_table(index=["source", "vae", "checkpoint"], columns="factor",
                                                             values="value")
```

```python
def plot_factor(factor):
    """Bars: reconstruction (hatched) and generated accuracy, coloured by VAE. Line: real test images."""
    df = accuracy[accuracy["factor"] == factor]
    rows = df[df["source"] != "real"].sort_values(["vae", "source"], ascending=[True, False])
    labels = rows["checkpoint"].fillna("reconstruction (" + rows["vae"] + ")")

    fig, ax = plt.subplots(figsize=(9, 8))
    for (vae, group), colour in zip(rows.groupby("vae", sort=False), plt.rcParams["axes.prop_cycle"].by_key()["color"]):
        bars = ax.bar(labels[group.index], group["value"], color=colour, label=vae)
        for bar, source in zip(bars, group["source"]):
            if source == "reconstruction":
                bar.set_hatch("//")
        ax.bar_label(bars, fmt="{:.4g}", padding=3, fontsize="small")

    real = df[df["source"] == "real"]["value"].mean()
    ax.axhline(real, color="black", linestyle="--", label=f"real ({real:.4g})")

    ax.set_title(f"{factor} accuracy (higher is better)")
    ax.tick_params(axis="x", labelrotation=45)
    plt.setp(ax.get_xticklabels(), ha="right")
    ax.legend(fontsize="small", loc="lower right")
    fig.tight_layout()
    plt.show()
```

```python
plot_factor("digit")
```

```python
plot_factor("fg")
```

```python
plot_factor("bg")
```
